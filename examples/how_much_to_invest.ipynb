```python
import numpy as np
import pandas as pd
from finance_calculators.calc_contributions import total_contribution, reformat, MaxContribution
```

```python
# Yearly income estimates using hourly rates and 52 pay periods.
# part-time, full-time, 4 hours overtime and 8 hours overtime
hours_per_week = np.array([15, 20, 30, 40, 50, 60])
hourly_rate = 75
weeks_in_a_year= 52
brute_income_yr= hours_per_week.reshape(-1,1) * hourly_rate * weeks_in_a_year
brute_income_yr
```

```python
brute_income_yr = np.array([50,75,100,150,200,250,500]).reshape(-1,1) * 1000
```

```python

year = 2026
max_contritions_dict = MaxContribution(year)
max_employee_contribution = max_contritions_dict.get("401k employee", year)

#This is an estimate. The max is 70000 - employee contribution for 2025
max_employer_contribution = max_contritions_dict.get("401k max", year) - max_employee_contribution

# Multiple investment percentages
percentage   = np.array([ .10, .12, .13, .14, .15, .16, .17, .18]).reshape(1,-1)
match_percentage  = np.array([.01, .02, .03, .05,.07, .10,.20, .25]).reshape(1,-1)

# Employee contribution per month
amounts_df = total_contribution(brute_income_yr,percentage,periods=12,max_contribution=max_employee_contribution)
amounts = reformat(amounts_df.copy(),periods=12,contribution_type= "employee")
display(amounts)

# Employer contribution per month
amounts_df_matching = total_contribution(brute_income_yr,match_percentage,periods=12,max_contribution=max_employer_contribution)
amounts = reformat(amounts_df_matching.copy(),periods=12,contribution_type= "employer")
display(amounts)
```

```python
# Employee contribution per year
amounts_df_yr = total_contribution(brute_income_yr,percentage,periods=1,max_contribution=max_employee_contribution)
amounts = reformat(amounts_df_yr.copy(),periods=1, contribution_type= "employee")
display(amounts)

# Employer contribution per year
amounts_df_yr_matching = total_contribution(brute_income_yr,match_percentage,periods=1,max_contribution=max_employer_contribution)
amounts = reformat(amounts_df_yr_matching.copy(),periods=1,contribution_type= "employer")
display(amounts)

```

```python
# Create a dataframe that adds the two contributions using one employer matching percentage
total_df = pd.DataFrame(amounts_df_yr.values + amounts_df_yr_matching.values,columns=amounts_df_yr.columns)
total_df.Income = amounts_df_yr.Income
columns = [f"{int(c)}/{int(b)}%" for b,c in zip(amounts_df_yr.columns.values[1:], amounts_df_yr_matching.columns.values[1:])]
total_df.columns = [amounts_df_yr.columns.values[0]] + columns
amounts = reformat(total_df.copy(),periods=1,contribution_type= "total")
display(amounts)
```
